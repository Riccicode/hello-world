1. Calculadora del índice de masa corporal

Escribe una función que calcule el IMC ingresado por el usuario/a, es decir, quien lo ejecute deberá ingresar estos datos. Puedes obtener más información de su cálculo en: 🔗 Índice de masa corporal IMC que es y cómo se calcula.

La función debe clasificar el resultado en sus respectivas categorías.

Consejo: Intenta validar los datos previamente, para que envíe un mensaje de advertencia si los datos introducidos por el usuario/a no están en el formato adecuado o no toma valores razonables.

```python
def calculoImc():
   #he colocado un while true para que en caso de error el programa no se cierre y pueda volver a ingresar los datos
    while True:
        try:
            peso = float(input("Ingrese su peso en kg, ej: 80.5 "))
            if peso <= 0:
                print("El peso no puede ser menor o igual a 0")
                continue
            break
        except ValueError: 
                print("Error, ha introducido un valor invalido, intentelo nuevamente")
               
        
          
    while True:
        try:
            altura = float(input("Ingrese su altura en metros, ej: 1.88"))
            if altura <= 0:
                print("La altura no puede ser menor o igual a 0")
                continue
            break
        except ValueError:
            print("Error, ha introducido un valor invalido, intentelo nuevamente")

    #calculo del imc y redondeo a 2 decimales

    IMC = round(peso / (altura ** 2),2)
    
    #funcion para determinar la categoria del imc
   
    def categoria(IMC):
        if IMC < 18.5:
         return "Bajo peso"
        elif 18.5 <= IMC < 25:
          return "Peso normal"
        elif 25 <= IMC < 30:
         return "Sobrepeso"
        else:
          return "Obesidad"
    
    print(f"Tu IMC es {IMC}. Clasificación: {categoria(IMC)}")
    
 

calculoImc()
```

2. Conversor de temperaturas

Existen diversas unidades de temperatura utilizadas en distintos contextos y regiones. Las más comunes son Celsius (°C), Fahrenheit (°F) y Kelvin (K). También existen otras unidades como Rankine (°Ra) y Réaumur (°Re).

Selecciona al menos 2 conversores, de modo que al introducir una temperatura devuelva, como mínimo, dos conversiones, de modo que se puedan guardar (recuerda que un print() no se puede guardar nunca).

Consejo: Intenta validar los datos previamente, para que envíe un mensaje de advertencia si los datos introducidos por el usuario/a no están en el formato adecuado.

(EXTRA): Piensa una manera de almacenar todas las posibles conversiones en un solo objeto (¿Lista? ¿Diccionario? ¿DataFrame?) en vez de escribir muchos if else en función de la temperatura de origen y la temperatura de destino.

```python
#defino la funcion para convertir temperaturas, con validacion de datos y manejo de errores

def convertir_temperatura():
    while True:
        try:
            valor = float(input("Introduce la temperatura: "))
            break
        except ValueError:
            print(" El valor introducido no es numérico.")
    while True:
       
            unidad_origen = str(input("Introduce la unidad (C, F o K): ")).upper()
            if unidad_origen in ["C", "F", "K"]:
                break
            else:
                print(" Unidad no válida. Usa C, F o K.")

#CREO UN DICCIONARIO PARA LAS CONVERSIONES DE TEMPERATURA, UTILIZANDO LAMBDAS PARA CALCULAR LAS CONVERSIONES DE MANERA DINAMICA
    conversiones = {
        "C": {
            "F": lambda x: x * 9/5 + 32,
            "K": lambda x: x + 273.15
        },
        "F": {
            "C": lambda x: (x - 32) * 5/9,
            "K": lambda x: (x - 32) * 5/9 + 273.15
        },
        "K": {
            "C": lambda x: x - 273.15,
            "F": lambda x: (x - 273.15) * 9/5 + 32
        }
    }

#hago un for para recorrer el diccionario y buscar la unidad de orgine y asignarle el resultado de la conversion
    resultadosConversiones = {}

    for unidad_destino, funcion in conversiones[unidad_origen].items():
        resultadosConversiones[unidad_destino] = round(funcion(valor), 2)
    
    
    #la funcion devuelve los valores de la temperatura original, la unidad de origen y un diccionario con las conversiones a las otras unidades
    return valor, unidad_origen, resultadosConversiones


valor, unidad_origen, resultadosConversiones = convertir_temperatura()

print("Temperatura:", valor, "°", unidad_origen, 
      ": Temperatura en otras regiones", resultadosConversiones)


```

3. Contador de palabras de un texto.

Escribe una función que dado un texto, muestre las veces que aparece cada palabra. Intenta que se gestionen todas las posibles casuísticas que hagan que el programa no funcione correctamente.

(EXTRA): ¿Cuál es la longitud media de las palabras del texto que has escrito? “¿Hola cómo va?” debería devolver (4+4+2) / 3 = 3.33

```python
import re
#utilizo re para eliminar los signos de puntuacion y contar solo las palabras, ademas de convertir todo a minusculas para evitar contar la misma palabra con mayusculas como diferente
#creo funcion para contar las palabras
def contador_de_palabras():
    texto = input("Introduce un texto: ")
    palabras = texto.split()
    palabras = re.findall(r'\b\w+\b', texto.lower())   #divido el texto en palabras y las palabras elimino todos los signos y convierto a minusculas para evitar contar la misma palabra con mayusculas como diferente
    conteo = {}
    
    #hago un for para recorrer la lista de palabras y contar cuantas veces aparece cada palabra, guardando el resultado en un diccionario conteo{palabra: cantidad}
    for palabra in palabras:
                   
        if palabra in conteo:
            conteo[palabra] += 1
        else:
            conteo[palabra] = 1
 
    total_longitud = sum(len(palabra) for palabra in palabras) #calculo la longitud total de todas las palabras sumando la longitud de cada palabra
    longitud_media = total_longitud / len(palabras) #calculo la longitud media dividiendo la longitud total entre el numero de palabras
    
    #imprimo dentro de la funcion
    print(texto)
    print("\nConteo de palabras:")
    #por cada palabra y cantidad en conteo imprimo esa palabras y las cantidades, ademas de la longitud media de las palabras al final
    for palabra, cantidad in conteo.items():
        print(f"{palabra}: {cantidad}")
    print(f"Longitud media de las palabras: {longitud_media}")

contador_de_palabras()
```

4. Diccionario inverso (con posibilidad de duplicados)

Resulta que el cliente tiene una encuesta muy antigua que se almacena en un diccionario y los resultados los necesita a la inversa, es decir, intercambiando las claves y valores. Los valores y claves en el diccionario original son únicos; si éste no es el caso, la función debería imprimir un mensaje de advertencia, junto con una lista con los valores asociados a la clave repetida.

```python
#creo la funcion
def reverse_dictionary(diccionario):
    diccionario_inverso = {}
    duplicados = {}
           
     # recorrer el diccionario y agregar al diccionario inverso dando vuelta clave y valor y los coloco en los diccionarios creados
    for clave, valor in diccionario.items():
        if valor not in diccionario_inverso:
            diccionario_inverso[valor] = clave
        else:
            # Registrar duplicados (valor que ya existe en el diccionario inverso lo agrego a duplicados)
            if valor not in duplicados:
                duplicados[valor] = [diccionario_inverso[valor]]
            
            duplicados[valor].append(clave)

            # Convertir a lista los valores duplicados en el diccionario inverso 
            # para poder agregar multiples claves
            if not isinstance(diccionario_inverso[valor], list):
                diccionario_inverso[valor] = [diccionario_inverso[valor]]
            
            diccionario_inverso[valor].append(clave)

    # Mostrar advertencia si hay duplicados
    if duplicados:
        print("⚠ Advertencia: Hay valores repetidos en el diccionario original.")
        for valor, claves in duplicados.items():
            print(f"El valor '{valor}' está asociado a las claves: {claves}")
    #la fucnion devuelve el diccionario invertido
    return diccionario_inverso

reverse_dictionary({'x': 'apple', 'y': 'banana', 'z': 'banana'})
```


2. Conversión de tipos de datos.

El cliente recibe una lista de datos y necesita generar dos listas, la primera donde estarán todos los elementos que pudieron convertirse en flotantes y la otra donde están los elementos que no pudieron convertirse. Ejemplo de la lista que recibe el cliente:
[ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']]



```python
#lista de datos del cliente
datos = [
    '1.3',
    'one',
    '1e10',
    'seven',
    '3-1/2',
    ('2', 1, 1.4, 'not-a-number'),
    [1, 2, '3', '3.4']
]
#creo dos listas vacias para guardar los valores convertibles a float y los no convertibles
convertibles = []
no_convertibles = []

# para cada elemento de la lista datos hago un for if para verificar primero si es list o tuple, los recorro y veo si pueden convertirce a float

for elemento in datos:

    # Si es lista o tupla → recorrer interior
    if isinstance(elemento, (list, tuple)):
        for subelemento in elemento:
            try:
                valor = float(subelemento)
                convertibles.append(valor)
            except ValueError:
                no_convertibles.append(subelemento)

    # Si es elemento  simple intento convertir a float y si no se puede lo agrego a no convertibles
    else:
        try:
            valor = float(elemento)
            convertibles.append(valor)
        except ValueError:
            no_convertibles.append(elemento)

#los imprimo

print("Convertibles:")
print(convertibles)

print("\nNo convertibles:")
print(no_convertibles)
```

En este punto, el cliente ha detectado un problema con las contraseñas que utilizan sus trabajadores. Asdf1234, fechas de cumpleaños o similares. Para solucionarlo, nos ha encargado una función de Python que genere contraseñas más seguras. La función debe depender de los siguientes parámetros:

```python

#importo numpy para generar numeros aleatorios y string para obtener los caracteres necesarios para generar la contraseña
import numpy as np
import string
#creo la funicon con los valores por defecto.
def generar_contraseña(
    longitud: int,
    mayusculas: bool = True,
    minusculas: bool = True,
    numeros: bool = True,
    signos: bool = False
                        ):
    #verifico la longitud y que se haya activado al menos un tipo de caracter, si no se cumple alguna de estas condiciones lanzo un error
    if longitud <= 0:
        raise ValueError("La longitud debe ser mayor que 0.")
    
    caracteres = ""
    obligatorios = ""
    #si los valores son true, realiza el if. cada uno asigna a la variable obligatorios un caracter aleatorio y agrego a la variable caracteres todos los caracteres para luego utlizarlos
    if mayusculas:
        obligatorios += "".join(np.random.choice(list(string.ascii_uppercase), 1))
        caracteres += string.ascii_uppercase
    
    if minusculas:
        obligatorios += "".join(np.random.choice(list(string.ascii_lowercase), 1))
        caracteres += string.ascii_lowercase
    
    if numeros:
        obligatorios += "".join(np.random.choice(list(string.digits), 1))
        caracteres += string.digits
    
    if signos:
        obligatorios += "".join(np.random.choice(list(string.punctuation), 1))
        caracteres += string.punctuation
    if not caracteres:
        raise ValueError("Debe activarse al menos un tipo de carácter.") # si no hay ningun true en los parametros te lo solicita obligatoriamente.
    
    # Convertimos a lista para numpy
    caracteres_array = list(caracteres)
    obligatorios_array = list(obligatorios)
    #genero la contraseña uniendo los caracteres obligatorios y luego agregando caracteres aleatorios hasta completar la longitud deseada
    contraseña = "".join(obligatorios_array) + "".join(
        np.random.choice(caracteres_array, size=(longitud-len(obligatorios_array)))) #la longitud es la que dan en el parametro - la cantidad que hay en obligatorios.

    #para que no se vean los caracteres obligatorios al principio de la contraseña, mezclo la contraseña generada con shuffle
    contraseña_lista = list(contraseña)
    np.random.shuffle(contraseña_lista)
    contraseña = "".join(contraseña_lista)

    print(obligatorios)
    return contraseña

generar_contraseña(10,True,True,True,False)
```

```python
import numpy as np
import string

def generar_contraseña(
    longitud: int,
    mayusculas: bool = True,
    minusculas: bool = True,
    numeros: bool = True,
    signos: bool = False
):

    if longitud <= 0:
        raise ValueError("La longitud debe ser mayor que 0.")

    caracteres = ""
    obligatorios = []

    if mayusculas:
        obligatorios.append(np.random.choice(list(string.ascii_uppercase)))
        caracteres += string.ascii_uppercase

    if minusculas:
        obligatorios.append(np.random.choice(list(string.ascii_lowercase)))
        caracteres += string.ascii_lowercase

    if numeros:
        obligatorios.append(np.random.choice(list(string.digits)))
        caracteres += string.digits

    if signos:
        obligatorios.append(np.random.choice(list(string.punctuation)))
        caracteres += string.punctuation

    if not caracteres:
        raise ValueError("Debe activarse al menos un tipo de carácter.")

    if longitud < len(obligatorios):
        raise ValueError("Longitud insuficiente para los tipos obligatorios.")

    restantes = longitud - len(obligatorios)

    adicionales = list(np.random.choice(list(caracteres), size=restantes))

    resultado = obligatorios + adicionales

    np.random.shuffle(resultado)
    print(caracteres)
    print(obligatorios)
    return "".join(resultado)

generar_contraseña(10,True,True,True,True)
```
