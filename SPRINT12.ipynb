1. Consulta l’API pública JSONPlaceholder utilitzant el mètode GET per obtenir:

Llista de publicacions (/posts)
Llista d’usuaris (/users)
Llista de tasques (/todos)

```python
import requests
import pandas as pd

URL = "https://jsonplaceholder.typicode.com"
```

```python
link_posts = requests.get(URL + "/posts")   # Hacemos la petición a la API para obtener los posts

posts = link_posts.json() # Convertimos la respuesta a formato JSON para poder trabajar con ella

df_posts = pd.DataFrame(posts) 

df_posts.head()
```

```python
link_users = requests.get(URL + "/users")

users = link_users.json()

df_users = pd.DataFrame(users)

df_users.head()
```

```python
link_todos = requests.get(URL + "/todos")

todos = link_todos.json()

df_todos = pd.DataFrame(todos)

df_todos.head()
```

```python

print("POSTS")
print("Cantidad:", len(df_posts))
print("Status code:", link_posts.status_code)
print("-" * 30)

print("USERS")
print("Cantidad:", len(df_users))
print("Status code:", link_users.status_code)
print("-" * 30)

print("TODOS")
print("Cantidad:", len(df_todos))
print("Status code:", link_todos.status_code)
```

4. Fes una petició POST per crear una nova publicació fictícia. Inclou el títol, el cos del missatge i un userId.

Mostra la resposta JSON.
Mostra el codi d'estat.

```python




# Datos de la nueva publicacion
new_post = {
    "title": "Analisis de datos",
    "body": "Este es un post de prueba para practicar requests en APIs",
    "userId": 1
}

# Peticion
response = requests.post(URL + "/posts", json=new_post)

# Mostrar resultados
print("Status code:", response.status_code)
print("Respuesta JSON:")
print(response.json())
```

5. Fes una petició PATCH per modificar parcialment una publicació existent.

Mostra la resposta JSON.
Mostra el codi d'estat.

```python

#PUT reemplaza todo y PATCH solo el campo que se indique


# ID de publicación existente
post_id = 1

# Datos a modificar (solo un campo)
update_data = {
    "title": "Nuevo titulo actualizado ",
    "body": "Contenido actualizado del post"
}

# Petición PATCH
response = requests.patch(URL + f"/posts/{post_id}", json=update_data)

# Resultados
print("Status code:", response.status_code)
print("Respuesta JSON:")
print(response.json())
```

6. Fes una petició DELETE sobre una publicació.

Mostra la resposta JSON.
Mostra el codi d'estat.

```python
#  DELETE
response = requests.delete(URL + f"/posts/{post_id}")

# Resultados
print("Status code:", response.status_code)
print("Respuesta JSON:")
print(response.json())
```


3. Fes una petició GET senzilla:

Mostra el codi d'estat de la resposta.
Imprimeix de forma clara alguns dels camps de la resposta JSON.

```python
import requests

url = "https://v3.football.api-sports.io/teams"

headers = {
    "x-apisports-key": "511021b484b52f1abe2bbda180f2a400"
}

params = {
    "country": "Argentina"
}

response = requests.get(url, headers=headers, params=params)

print("Status code:", response.status_code)

data_futbol = response.json()

# VALIDACION
if len(data_futbol["response"]) > 0:
    print("Total equipos:", len(data_futbol["response"]))
    print("-" * 40)
    
    # Mostrar hasta 10 equipos
    for item in data_futbol["response"][:10]:
        team = item["team"]
        venue = item["venue"]
        
        # TEAM
        print("id:", team["id"])
        print("name:", team["name"])
        print("code:", team.get("code"))
        print("country:", team["country"])
        print("founded:", team["founded"])
        print("national:", team["national"])
        print("logo:", team["logo"])
        
        # VENUE
        print("stadium:", venue["name"])
        print("address:", venue["address"])
        print("city:", venue["city"])
        print("capacity:", venue["capacity"])
        print("surface:", venue["surface"])
        
        print("-" * 40)
else:
    print("No hay datos")
```

```python
import pandas as pd

teams_list = []

for item in data_futbol["response"]:
    
    team = item["team"]
    venue = item["venue"]
    
    teams_list.append({
        "team_id": team["id"],
        "team_name": team["name"],
        "code": team["code"],
        "country": team["country"],
        "founded": team["founded"],
        "national": team["national"],
        "stadium": venue["name"],
        "address": venue["address"],
        "logo": team["logo"],
        "city": venue["city"],
        "capacity": venue["capacity"],
        "surface": venue["surface"]
    })

# Crear DataFrame
df_teams = pd.DataFrame(teams_list)

# Mostrar primeras filas
df_teams.to_csv("teams_argentina.csv", index=False) 

df_teams
```

Nivell 3
API d'Open Data Barcelona
1. Utilitza l’API d’Open Data BCN per:

Cercar un dataset del teu interès mitjançant package_search i package_show. .


```python
import requests


busca_placas = "plaques fotovoltaiques"

base = "https://opendata-ajuntament.barcelona.cat/data/api/3/action"

search = requests.get(f"{base}/package_search?q={busca_placas}").json()

# Mostro el resultat amb el package_show, llista tots els títols inclosos
for busqueda in search["result"]["results"]:
    print("Nombre:", busqueda["title"])
    print("RESOURCE_ID:", busqueda["id"])
    print("-" * 50)
```

2. Dels resultats obtinguts, selecciona’n un que tingui recursos disponibles en CSV o JSON.

```python

dataset_placas = search["result"]["results"][0]    

for resource in dataset_placas["resources"]:
    print("Nombre:", resource["name"])
    print("Formato:", resource["format"])
    print("URL:", resource["url"])
    print("RESOURCE_ID:", resource["id"])
    print("-" * 50)
```

3. Amb el resource_id del recurs seleccionat, realitza una consulta mitjançant datastore_search per:

Recuperar almenys 100 registres.

```python
resource_id = dataset_placas["resources"][0]["id"]

data_placas = requests.get(
    f"{base}/datastore_search?resource_id={resource_id}&limit=100"
).json()
```

4. Converteix els resultats a un DataFrame.




```python
import pandas as pd
data_placas_df = pd.DataFrame(data_placas["result"]["records"])

data_placas_df

```

5. Desa el DataFrame en un fitxer .csv.

```python
data_placas_df.to_csv("placasfotovoltaiquesbarcelona.csv", index=False) 
```
